```python
from pathlib import Path
from collections import Counter
import csv

file_to_load = Path('election_results.csv')
candidate_votes = Counter()
county_votes = Counter()
with file_to_load.open(newline='', encoding='utf-8-sig') as election_data:
    reader = csv.DictReader(election_data)
    if not {'County', 'Candidate'}.issubset(reader.fieldnames or []):
        raise ValueError('The input must include County and Candidate columns.')
    for row in reader:
        candidate_votes[row['Candidate']] += 1
        county_votes[row['County']] += 1
total_votes = sum(candidate_votes.values())
if total_votes == 0:
    raise ValueError('The input contains no votes.')
```

```python
lines = ['Election Results', '-------------------------', f'Total Votes: {total_votes:,}', '-------------------------', 'County Votes:']
for county, votes in county_votes.items():
    lines.append(f'{county}: {votes / total_votes:.1%} ({votes:,})')
largest_turnout = max(county_votes, key=county_votes.get)
lines.extend(['-------------------------', f'Largest County Turnout: {largest_turnout}', '-------------------------'])
for candidate, votes in candidate_votes.items():
    lines.append(f'{candidate}: {votes / total_votes:.1%} ({votes:,})')
winning_count = max(candidate_votes.values())
winners = [name for name, count in candidate_votes.items() if count == winning_count]
lines.extend(['-------------------------', f"Winner: {', '.join(winners)}", f'Winning Vote Count: {winning_count:,}', f'Winning Percentage: {winning_count / total_votes:.1%}', '-------------------------'])
report = '\n'.join(lines) + '\n'
print(report)
output_dir = Path('analysis')
output_dir.mkdir(exist_ok=True)
(output_dir / 'election_analysis.txt').write_text(report, encoding='utf-8')
```
